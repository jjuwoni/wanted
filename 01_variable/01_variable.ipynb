## 01. Python 변수
* 변수란? 값을 담을 수 있는 공간이다.
* Java 와는 다르게 Python 은 변수의 타입을 선언 시에 지정하지 않는다.
* Python 은 런타임 시점에서 값을 보고 타입을 판단하게 된다.

```python
user_question = "python 이 뭐야?"

print(user_question)
```

```python
model_name = "stt-ai-model"
max_token = 1000
temperature = 0.7
is_streaming = True

print(model_name)
print(max_token)
print(temperature)
print(is_streaming)
```

## Python 의 자료형은 런타임 시 결정된다.
* type() 함수로 변수의 타입을 확인할 수 있다.

```python
print(type(model_name))
print(type(max_token))
print(type(temperature))
print(type(is_streaming))
```

```python
name = "raccoon"
user_question = "RAG 가 뭐야?"

# raccoon 의 질문 : RAG 가 뭐야?

print(name + "의 질문 : " + user_question)

# f-string

message = f"{name}의 질문 : {user_question}"

print(message)
```
