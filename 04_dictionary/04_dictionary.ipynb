# 04. Dictionary
* Dictionary 는 Java 에서의 Map 과 비슷한 역할을 하게 된다.
* Map<String , Object> 형태라고 생각하는 것이 좋다.
* Json : { name : "조평훈" }
* Key - Value 형태의 자료구조이다!!!!!!!!!!!

```python
user = {
    "name" : "raccoon",
    "age" : 20 , 
    "course" : "AI 활용 Backend 과정"
}

print(user)
print(type(user))
```

```python
# dictionary 타입의 값을 꺼내오는 방법
print(user["name"]) # ["key"] -> value
print(user.get("age")) # get("key") -> value
```

```python
# dictionary 타입의 값을 수정하는 방법
user["name"] = "koala"

# dictionary 타입의 값을 삭제하는 방법
del user["age"]

print(user)
```

```python
# Class 와 함께 Dictionary 자료형 활용해보기
# 1~4 총 집합

# AI 응답 관련 Service Class
class AIService:

    def answer(self , request : dict) -> dict :
        return {
            "question" : request["question"],
            "model" : request["model"],
            "answer" : f"{request.get("question")} 에 대한 답변!"
        }

# AI 응답 객체 생성
service = AIService()

# AI 응답 요청 해보기
response = service.answer({
    "question" : "FastAPI 가 도대체 뭐야?",
    "model" : "gemini-slim"
})

print(response)
```
