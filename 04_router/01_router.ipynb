# 04. Router

* FastAPI 의 APIRouter 를 활용해서 API 관련 코드를 분리하는 방법
* 문제 상황
    * main.py 파일에 모든 API 및 DTO 를 작성해두었다.
    * 프로젝트의 규모 및 기능이 추가 될 수록 main.py 복잡해진다.

* 필요 사항
    * Router 의 개념이 필요하게 된다.
    * Spring Controller == FastAPI Router
    * APIRouter 사용해서 main.py 파일에 Router 를 등록한다.

* main.py
    * FastAPI 객체 생성
    * Router 등록
* chat_router.py
    * ChatBot 관련 기능 API
* rag_router.py
    * RAG 관련 기능 API
