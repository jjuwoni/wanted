users = [

* 하나의 데이터만을 다루는 경우보다 , 여러 개의 데이터를 다루는 경우가 훨씬 많다.

````python
# Python List 기본 형태

# Java [] : 배열
# 배열의 특징 : 크기 정해져야함. 순서가 있다. 자료형이 같아야 한다.
 
 # Python [] : List
 # Java 의 배열과 가장 큰 차이점 : 자료형은 상관이 없다.
messages = [
    "안녕",
    20,
    "raccoon"
] 

print(messages)
print(type(messages))
````

````python
messages = [
    "안녕",
    20,
    "raccoon"
] 

# 데이터에 접근하는 방법
# 배열의 특성을 떠올려보자. Index 번호 체계에 의한 순서가 존재한다.!!

print(messages[0]) # 변수명[index]

# 데이터를 추가하는 방법
# List<UserDTO> users = new ArrayList<>;
# 4번 회원을 추가하고 싶다.
# users.add
# append() 메서드가 Java 진영에서의 add() 와 동일한 역할을 하게 된다.

messages.append("koala")

print(messages)

# 데이터를 수정하는 방법
messages[3] = "tiger"

print(messages)

# 데이터를 삭제하는 방법
# remove() , pop()

messages.remove("tiger")
messages.pop(0)

print(messages)

# 배열의 크기를 알아보는 메서드
# java -> size()
# python -> len() length 의 축약어

print(len(messages))

# Python 에서의 반복문
messages2 = ["java" , "js" , "python" , "c#" , "c"]

for message in messages2:
    print(message)
    
````

```java
// 향상 된 for 문
// Python 에서의 반복문은 Java 에서 향상된 for 문과 비슷한 구조를
// 가지고 있게 된다.
for(String message : messages) {
    System.out.println(message);
}

```

````python
# 지금까지 배운 개념 누적 적용

class AIService:

    def answer(self, question : str) -> dict :
        return {
            "question" : question,
            "answer" : f"{question} 에 대한 답변입니다!"
        }

ai_service = AIService()

questions = [
    "Python 이 뭐임?",
    "점심 뭐 먹지?",
    "FastAPI 가 뭐에요 도대체?"
]

responses = []

for question in questions:
    response = ai_service.answer(question)
    responses.append(response)

print(responses)

````
